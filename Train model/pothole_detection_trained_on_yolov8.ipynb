```python
from ultralytics import YOLO

# Load a YOLOv8 model (choose s, m, l depending on speed/accuracy)
model = YOLO("yolov8m.pt")  # or yolov8s.pt, yolov8m.pt, yolov8l.pt

# Train
model.train(
    data="/content/Pothole-Detection--10/data.yaml",   # path to data.yaml
    epochs=250,                  # training epochs
    imgsz=640,                  # image size
    batch=16,                   # batch size
    name="pothole_model"        # experiment name
)
```

```python
from ultralytics import YOLO

# Load trained YOLOv8 model
model = YOLO("/content/runs/detect/train/weights/best.pt")

# Run prediction on a video and save results
results = model.predict(
    source="/content/1000256987.mp4",   # path to video or 0 for webcam
    show=True,                  # show predictions live
    save=True,                  # save predictions to "runs/detect/predict"
    conf=0.5
)

print("Predictions saved in:", results[0].save_dir)
```
